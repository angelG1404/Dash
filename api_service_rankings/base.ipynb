```python
import sqlite3

from fastapi import FastAPI

app = FastAPI()


def get_connection():
    conn = sqlite3.connect("BD_rankings_universidades.db")
    conn.row_factory = sqlite3.Row
    return conn


@app.get("/universidades")
def obtener_universidades():
    conn = get_connection()

    rows = conn.execute(
        """
        select r.nombre_ranking, s.area_grande, s.nombre_espanol
        from dim_ranking as r
        join dim_subject as s on r.id_ranking = s.id_ranking
        where r.id_ranking = ?
        """, ('id_ranking',)
    ).fetchall()

    conn.close()

    return [dict(row) for row in rows]
```
